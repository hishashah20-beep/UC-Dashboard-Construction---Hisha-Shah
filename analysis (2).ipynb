## Load data

```python
import pandas as pd

df = pd.read_csv("bay_area_modeling_table.csv", low_memory=False)
```

## Filter to Fall 2025, Bay Area campuses

We exclude "Universitywide" — that row counts distinct students admitted to
any UC campus, not per-campus application counts, so including it would
double-count and distort the comparison (see README for detail).

```python
y = df[(df.fall_term == 2025) & (df.campus != "Universitywide")]
```

## Aggregate by campus

We sum applicants and admits within each campus first, then divide —
**not** average the per-school admit rate — so large and small schools
are weighted by applicant volume, not counted equally.

```python
campus_summary = (
    y.groupby("campus")
     .agg(total_applicants=("applicants", "sum"),
          total_admits=("admits", "sum"))
     .assign(admit_rate=lambda d: d.total_admits / d.total_applicants)
     .reset_index()
     .sort_values("admit_rate")
)

print(campus_summary)
```

## Visualize

```python
import matplotlib.pyplot as plt

plt.barh(campus_summary.campus, campus_summary.admit_rate)
plt.xlabel("UC admit rate")
plt.title("UC admit rate by campus, Bay Area applicants, fall 2025")
plt.tight_layout()
plt.show()
```
