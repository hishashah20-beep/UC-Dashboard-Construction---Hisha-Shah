```python
import pandas as pd

df = pd.read_csv("bay_area_modeling_table.csv", low_memory=False)

# Fall 2025 only, exclude "Universitywide" (that's a different metric, see README)
y = df[(df.fall_term == 2025) & (df.campus != "Universitywide")]

# Sum counts per campus, then divide (not average)
campus_summary = (
    y.groupby("campus")
     .apply(lambda g: pd.Series({
         "total_applicants": g.applicants.sum(),
         "total_admits": g.admits.sum(),
         "admit_rate": g.admits.sum() / g.applicants.sum()
     }))
     .reset_index()
     .sort_values("admit_rate")
)

print(campus_summary)

import matplotlib.pyplot as plt
plt.barh(campus_summary.campus, campus_summary.admit_rate)
plt.xlabel("UC admit rate")
plt.title("UC admit rate by campus, Bay Area applicants, fall 2025")
plt.tight_layout()
plt.show()
```
